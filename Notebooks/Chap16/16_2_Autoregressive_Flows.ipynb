First we'll define an invertible one dimensional function as in figure 16.5

```python
import numpy as np
# First let's make the 1D piecewise linear mapping as illustrated in figure 16.5
def g(h, phi):
  # Equation 16.12 implements a piecewise linear, monotonic function.
  # The 'phi' vector contains 5 elements that sum to 1, representing the heights
  # of the output 'h_prime' at equally spaced input intervals.
  # Specifically, phi[k] is the change in h_prime for the k-th segment of h.

  N = len(phi)
  # Calculate cumulative sum of phi, starting with 0
  phi_cumulative = np.cumsum(phi)
  phi_cumulative = np.insert(phi_cumulative, 0, 0) # [0, phi[0], phi[0]+phi[1], ...]

  # Determine which segment 'h' falls into
  # Input intervals are [k/N, (k+1)/N]
  # k will be an integer from 0 to N-1
  k = int(np.floor(h * N))
  k = np.clip(k, 0, N - 1) # Ensure k is within valid segment indices

  # Get the lower bound of the input segment
  h_lower_bound = k / N

  # Get the lower bound of the output segment
  y_lower_bound = phi_cumulative[k]

  # The slope of the current segment is phi[k] / (1/N) = phi[k] * N
  # h_prime = y_lower_bound + slope * (h - h_lower_bound)
  h_prime = y_lower_bound + phi[k] * N * (h - h_lower_bound)

  return h_prime
```

```python
# Let's test this out.  If you managed to vectorize the routine above, then good for you
# but I'll assume you didn't and so we'll use a loop

# Define the parameters
phi = np.array([0.2, 0.1, 0.4, 0.05, 0.25])

# Run the function on an array
h = np.arange(0,1,0.01)
h_prime = np.zeros_like(h)
for i in range(len(h)):
  h_prime[i] = g(h[i], phi)

# Draw the function
fig, ax = plt.subplots()
ax.plot(h,h_prime, 'b-')
ax.set_xlim([0,1])
ax.set_ylim([0,1])
ax.set_xlabel('Input, $h$')
ax.set_ylabel('Output, $h^\prime$')
plt.show()
```

We will also need the inverse of this function

```python
# Define the inverse function
def g_inverse(h_prime, phi):
    # Lot's of ways to do this, but we'll just do it by bracketing
    h_low = 0
    h_mid = 0.5
    h_high = 0.999

    thresh = 0.0001
    c_iter = 0
    while(c_iter < 20 and h_high - h_low > thresh):
        h_prime_low = g(h_low, phi)
        h_prime_mid = g(h_mid, phi)
        h_prime_high = g(h_high, phi)
        if h_prime_mid < h_prime:
          h_low = h_mid
        else:
          h_high = h_mid

        h_mid = h_low+(h_high-h_low)/2
        c_iter+=1

    return h_mid
```

Now let's define an autoregressive flow.  Let's switch to looking at figure 16.7.# We'll assume that our piecewise function will use five parameters phi1,phi2,phi3,phi4,phi5

```python
def ReLU(preactivation):
  activation = preactivation.clip(0.0)
  return activation

def softmax(x):
  x = np.exp(x) ;
  x = x/ np.sum(x) ;
  return x

# Return value of phi that doesn't depend on any of the inputs
def get_phi():
  return np.array([0.2, 0.1, 0.4, 0.05, 0.25])

# Compute values of phi that depend on h1
def shallow_network_phi_h1(h1, n_hidden=10):
  # For neatness of code, we'll just define the parameters in the network definition itself
  n_input = 1
  np.random.seed(n_input)
  beta0 = np.random.normal(size=(n_hidden,1))
  Omega0 = np.random.normal(size=(n_hidden, n_input))
  beta1 = np.random.normal(size=(5,1))
  Omega1 = np.random.normal(size=(5, n_hidden))
  input_vector = np.array([float(h1)]).reshape(-1, 1) # Ensure input is a column vector
  return softmax(beta1 + Omega1 @ ReLU(beta0 + Omega0 @ input_vector))

# Compute values of phi that depend on h1 and h2
def shallow_network_phi_h1h2(h1,h2,n_hidden=10):
  # For neatness of code, we'll just define the parameters in the network definition itself
  n_input = 2
  np.random.seed(n_input)
  beta0 = np.random.normal(size=(n_hidden,1))
  Omega0 = np.random.normal(size=(n_hidden, n_input))
  beta1 = np.random.normal(size=(5,1))
  Omega1 = np.random.normal(size=(5, n_hidden))
  input_vector = np.array([float(h1), float(h2)]).reshape(-1, 1) # Ensure input is a column vector
  return softmax(beta1 + Omega1 @ ReLU(beta0 + Omega0 @ input_vector))

# Compute values of phi that depend on h1, h2, and h3
def shallow_network_phi_h1h2h3(h1,h2,h3, n_hidden=10):
  # For neatness of code, we'll just define the parameters in the network definition itself
  n_input = 3
  np.random.seed(n_input)
  beta0 = np.random.normal(size=(n_hidden,1))
  Omega0 = np.random.normal(size=(n_hidden, n_input))
  beta1 = np.random.normal(size=(5,1))
  Omega1 = np.random.normal(size=(5, n_hidden))
  input_vector = np.array([float(h1), float(h2), float(h3)]).reshape(-1, 1) # Ensure input is a column vector
  return softmax(beta1 + Omega1 @ ReLU(beta0 + Omega0 @ input_vector))
```

The forward mapping as shown in figure 16.7 a

```python
def forward_mapping(h1,h2,h3,h4):
  # h_prime1 depends only on h1 and a fixed phi
  phi1 = get_phi()
  h_prime1 = g(h1, phi1)

  # h_prime2 depends on h2 and phi from h_prime1
  phi2 = shallow_network_phi_h1(h_prime1)
  h_prime2 = g(h2, phi2)

  # h_prime3 depends on h3 and phi from h_prime1, h_prime2
  phi3 = shallow_network_phi_h1h2(h_prime1, h_prime2)
  h_prime3 = g(h3, phi3)

  # h_prime4 depends on h4 and phi from h_prime1, h_prime2, h_prime3
  phi4 = shallow_network_phi_h1h2h3(h_prime1, h_prime2, h_prime3)
  h_prime4 = g(h4, phi4)

  return h_prime1, h_prime2, h_prime3, h_prime4
```

The backward mapping as shown in figure 16.7b

```python
def backward_mapping(h1_prime,h2_prime,h3_prime,h4_prime):
  # h1 depends on h1_prime and a fixed phi
  phi1 = get_phi()
  h1 = g_inverse(h1_prime, phi1)

  # h2 depends on h2_prime and phi from h1_prime
  phi2 = shallow_network_phi_h1(h1_prime)
  h2 = g_inverse(h2_prime, phi2)

  # h3 depends on h3_prime and phi from h1_prime, h2_prime
  phi3 = shallow_network_phi_h1h2(h1_prime, h2_prime)
  h3 = g_inverse(h3_prime, phi3)

  # h4 depends on h4_prime and phi from h1_prime, h2_prime, h3_prime
  phi4 = shallow_network_phi_h1h2h3(h1_prime, h2_prime, h3_prime)
  h4 = g_inverse(h4_prime, phi4)

  return h1,h2,h3,h4
```

Finally, let's make sure that the network really can be inverted

```python
# Test the network to see if it does invert correctly
h1 = 0.22; h2 = 0.41; h3 = 0.83; h4 = 0.53
print("Original h values %3.3f,%3.3f,%3.3f,%3.3f"%(h1,h2,h3,h4))
h1_prime, h2_prime, h3_prime, h4_prime = forward_mapping(h1,h2,h3,h4)
print("h_prime values %3.3f,%3.3f,%3.3f,%3.3f"%(h1_prime,h2_prime,h3_prime,h4_prime))
h1,h2,h3,h4 =  backward_mapping(h1_prime,h2_prime,h3_prime,h4_prime)
print("Reconstructed h values %3.3f,%3.3f,%3.3f,%3.3f"%(h1,h2,h3,h4))
```
